```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df
```

```python
df['Age'] = df['Age'].round(1)
```

```python
df
```

```python
df[['Last name', 'Rest']] = df['Name'].str.split(', ', expand = True)
df
```

```python
df[['Title', 'First_Middle_Nmaes']] = df['Rest'].str.split(' ', n=1, expand = True)
df
```

```python
df.drop(columns = ['Name', 'Rest'], inplace = True)
df
```

```python
df['Sex'] = df['Sex'].str.replace('male', 'Male')
df['Sex'] = df['Sex'].str.replace('feMale', 'Female')

df
```

```python
df['Cabin'] = df['Cabin'].fillna('')
df
```

```python
df[['First Name', 'Middle_Nmaes']] = df['First_Middle_Nmaes'].str.split(' ', n=1, expand = True)
df
```

```python
df[['Middle Name', 'Other Name9s0']] = df['Middle_Nmaes'].str.split(' ', n=1, expand = True)
df
```

```python
df.drop(columns = ['First_Middle_Nmaes', 'Middle_Nmaes'], inplace = True)
df
```

```python
# New column order
new_column_order = ['PassengerId', 'Survived', 'Pclass', 'Title', 'First Name', 'Middle Name', 'Last name', 'Other Name9s0', 'Sex', 'Age', 'SibSp', 'Parch', 'Ticket', 'Fare', 'Cabin', 'Embarked']

df = df[new_column_order]
df
```

```python
df = df.rename(columns = {'First Name': 'FirstName', 'Middle Name': 'MiddleName', 'Last name': 'LastName', 'Other Name9s0': 'OtherName(s)'})
df
```

```python
df['OtherName(s)'] = df['OtherName(s)'].str.replace(r'[^a-zA-Z\s]', '', regex=True)
df
```
